````python
# ============================================================
# STEP 2A — ADD REMAINING 180 OPEN_FOOD_BUSINESS QUERIES
# ============================================================

import pandas as pd

cities = [
    "Mumbai", "Pune", "Thane", "Navi Mumbai",
    "Nagpur", "Nashik", "Kolhapur", "Solapur",
    "Amravati", "Kalyan", "Dombivli",
    "Chhatrapati Sambhajinagar"
]

business_types = [
    "restaurant",
    "bakery",
    "cafe",
    "cloud kitchen",
    "food stall",
    "juice centre",
    "snack centre",
    "small eatery",
    "food outlet",
    "tiffin service",
    "catering business",
    "sweet shop",
    "fast food shop",
    "tea cafe",
    "home kitchen"
]

templates = [
    "I want to start a {business} in {city}",
    "How can I open a {business} in {city}?",
    "What permissions are needed to start a {business} in {city}?",
    "{city} me {business} start karna hai",
    "{city} mein {business} kholna hai",
    "I am planning to open a {business} in {city}",
    "What do I need to start a {business} in {city}?",
    "Need guidance for opening a {business} in {city}",
    "Can I start a {business} in {city}?",
    "I need the process for opening a {business} in {city}",
    "What approvals do I need for a {business} in {city}?",
    "{city} me naya {business} shuru karna hai",
    "Mujhe {city} mein {business} open karna hai",
    "How do I legally start a {business} in {city}?",
    "I need help starting a {business} in {city}",
    "{city} mein {business} ke liye kya process hai?",
    "What documents are needed to open a {business} in {city}?",
    "I want to set up a {business} in {city}",
    "Maharashtra ke {city} mein {business} start karna hai",
    "Please tell me how to start a {business} in {city}",
    "What is the procedure to open a {business} in {city}?",
    "{business} business {city} me kaise start karu?",
    "I am thinking of opening a {business} in {city}",
    "{city} mein apna {business} shuru karna hai"
]

# Read existing CSV
existing_df = pd.read_csv(csv_path)

# Existing query texts so duplicates are not added
existing_texts = set(
    existing_df["text"].astype(str).str.strip().str.lower()
)

new_queries = []

# Generate diverse combinations
for b_index, business in enumerate(business_types):
    for c_index, city in enumerate(cities):

        # Rotate templates so every row doesn't follow the same pattern
        template_index = (b_index * 3 + c_index * 5) % len(templates)

        query = templates[template_index].format(
            business=business,
            city=city
        )

        normalized = query.strip().lower()

        if (
            normalized not in existing_texts
            and normalized not in {q.lower() for q in new_queries}
        ):
            new_queries.append(query)

# If any candidate collided with an existing query,
# generate extra unique variations until exactly 180 are reached.
if len(new_queries) < 180:

    for business in business_types:
        for city in cities:

            extra_query = (
                f"I need information about permissions and registration "
                f"for starting a {business} in {city}, Maharashtra"
            )

            normalized = extra_query.lower()

            if (
                normalized not in existing_texts
                and normalized not in {q.lower() for q in new_queries}
            ):
                new_queries.append(extra_query)

            if len(new_queries) == 180:
                break

        if len(new_queries) == 180:
            break

# Keep exactly 180
new_queries = new_queries[:180]

# Safety check
assert len(new_queries) == 180, \
    f"Expected 180 queries, got {len(new_queries)}"

# Create dataframe
food_df = pd.DataFrame({
    "text": new_queries,
    "intent": ["OPEN_FOOD_BUSINESS"] * 180
})

# Append to existing CSV
food_df.to_csv(
    csv_path,
    mode="a",
    header=False,
    index=False
)

print("New rows added:", len(food_df))
print("OPEN_FOOD_BUSINESS dataset should now have 200 rows.")
````

````python
unsupported_topics = [
    "passport renewal",
    "PAN card application",
    "Aadhaar update",
    "driving licence",
    "learner licence",
    "voter ID",
    "ration card",
    "income certificate",
    "caste certificate",
    "non creamy layer certificate",
    "marriage registration",
    "death certificate",
    "senior citizen certificate",
    "disability certificate",
    "vehicle ownership transfer"
]

cities = [
    "Mumbai", "Pune", "Thane", "Navi Mumbai",
    "Nagpur", "Nashik", "Kolhapur", "Solapur",
    "Amravati", "Kalyan", "Dombivli",
    "Chhatrapati Sambhajinagar"
]

templates = [
    "I need help with {topic} in {city}",
    "How can I apply for {topic} in {city}?",
    "{city} me {topic} ka process kya hai?",
    "I want information about {topic} in {city}",
    "What documents are required for {topic} in {city}?",
    "How do I get {topic} in Maharashtra?",
    "Mujhe {city} mein {topic} ke liye apply karna hai",
    "What is the procedure for {topic} in {city}?",
    "Need guidance for {topic} in {city}",
    "Where can I apply for {topic} in {city}?",
    "Please explain the process for {topic} in Maharashtra",
    "{topic} ke liye {city} me kya karna hoga?"
]

existing_df = pd.read_csv(csv_path)

existing_texts = set(
    existing_df["text"].astype(str).str.strip().str.lower()
)

new_unsupported_queries = []

for topic_index, topic in enumerate(unsupported_topics):
    for city_index, city in enumerate(cities):

        template_index = (topic_index * 3 + city_index * 2) % len(templates)

        query = templates[template_index].format(
            topic=topic,
            city=city
        )

        normalized = query.strip().lower()

        if (
            normalized not in existing_texts
            and normalized not in {q.lower() for q in new_unsupported_queries}
        ):
            new_unsupported_queries.append(query)

if len(new_unsupported_queries) < 180:
    for topic in unsupported_topics:
        for city in cities:

            extra_query = (
                f"I need detailed information about {topic} "
                f"in {city}, Maharashtra"
            )

            normalized = extra_query.lower()

            if (
                normalized not in existing_texts
                and normalized not in {q.lower() for q in new_unsupported_queries}
            ):
                new_unsupported_queries.append(extra_query)

            if len(new_unsupported_queries) == 180:
                break

        if len(new_unsupported_queries) == 180:
            break

new_unsupported_queries = new_unsupported_queries[:180]

assert len(new_unsupported_queries) == 180, \
    f"Expected 180 queries, got {len(new_unsupported_queries)}"

unsupported_more_df = pd.DataFrame({
    "text": new_unsupported_queries,
    "intent": ["OTHER_UNSUPPORTED"] * 180
})

unsupported_more_df.to_csv(
    csv_path,
    mode="a",
    header=False,
    index=False
)

print("New rows added:", len(unsupported_more_df))
print("OTHER_UNSUPPORTED dataset should now have 200 rows.")
````

````python
import pandas as pd

df = pd.read_csv(csv_path)

print("Shape:", df.shape)
print("\nClass counts:")
print(df["intent"].value_counts())
print("\nSample rows:")
print(df.sample(5, random_state=42))
````

````python
from sklearn.model_selection import train_test_split

X = df["text"]
y = df["intent"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining class counts:")
print(y_train.value_counts())

print("\nTesting class counts:")
print(y_test.value_counts())
````

````python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Train TF-IDF shape:", X_train_tfidf.shape)
print("Test TF-IDF shape:", X_test_tfidf.shape)
print("Vocabulary size:", len(vectorizer.vocabulary_))
````

````python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

model.fit(X_train_tfidf, y_train)

print("Model training complete.")
````

````python
from sklearn.metrics import accuracy_score, classification_report

y_pred = model.predict(X_test_tfidf)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
````

````python
DOMICILE_PURPOSES = {
    "admission": [
        "admission",
        "college",
        "university",
        "education"
    ],
    "scholarship": [
        "scholarship"
    ],
    "government_job": [
        "government job",
        "govt job",
        "job application"
    ],
    "state_benefit": [
        "state benefit",
        "government scheme",
        "scheme"
    ]
}

def extract_domicile_purpose(text):
    text_lower = text.lower()

    for purpose, keywords in DOMICILE_PURPOSES.items():
        for keyword in keywords:
            if keyword in text_lower:
                return purpose

    return None
````

````python
def extract_entities(text, predicted_intent):
    result = {
        "location": extract_location(text),
        "business_type": None,
        "event_type": None,
        "birth_subtype": None,
        "domicile_purpose": None
    }

    if predicted_intent == "OPEN_FOOD_BUSINESS":
        result["business_type"] = extract_business_type(text)

    elif predicted_intent == "ORGANIZE_PUBLIC_EVENT":
        result["event_type"] = extract_event_type(text)

    elif predicted_intent == "BIRTH_CERTIFICATE":
        result["birth_subtype"] = extract_birth_subtype(text)

    elif predicted_intent == "DOMICILE_CERTIFICATE_MAHARASHTRA":
        result["domicile_purpose"] = extract_domicile_purpose(text)

    return result
````

````python
MAHARASHTRA_CITIES = [
    "mumbai",
    "pune",
    "thane",
    "navi mumbai",
    "nagpur",
    "nashik",
    "kolhapur",
    "solapur",
    "amravati",
    "kalyan",
    "dombivli",
    "chhatrapati sambhajinagar"
]

OUTSIDE_MAHARASHTRA_CITIES = [
    "delhi",
    "new delhi",
    "bangalore",
    "bengaluru",
    "hyderabad",
    "chennai",
    "kolkata",
    "ahmedabad",
    "jaipur",
    "lucknow",
    "surat",
    "indore",
    "bhopal",
    "chandigarh",
    "patna",
    "ranchi",
    "kochi",
    "gurgaon",
    "gurugram",
    "noida"
]

def extract_location_with_scope(text):
    text_lower = text.lower()

    for city in MAHARASHTRA_CITIES:
        if city in text_lower:
            return {
                "location": city.title(),
                "location_scope": "MAHARASHTRA"
            }

    for city in OUTSIDE_MAHARASHTRA_CITIES:
        if city in text_lower:
            return {
                "location": city.title(),
                "location_scope": "OUTSIDE_MAHARASHTRA"
            }

    return {
        "location": None,
        "location_scope": "MISSING"
    }
````

````python
scope_tests = [
    "I want to open a cafe in Pune",
    "I want to open a cafe in Delhi",
    "birth certificate chahiye",
    "We want to organize an event in Bengaluru"
]

for query in scope_tests:
    print(query, "->", extract_location_with_scope(query))
````

````python
REQUIRED_FIELDS = {
    "OPEN_FOOD_BUSINESS": [
        "location",
        "business_type"
    ],

    "ORGANIZE_PUBLIC_EVENT": [
        "location",
        "event_type"
    ],

    "BIRTH_CERTIFICATE": [
        "location"
    ],

    "DOMICILE_CERTIFICATE_MAHARASHTRA": [],

    "OTHER_UNSUPPORTED": []
}

print(REQUIRED_FIELDS)
````

````python
def detect_missing_fields(intent, entities, location_scope):
    # Outside Maharashtra = out of scope, not "missing"
    if location_scope == "OUTSIDE_MAHARASHTRA":
        return {
            "status": "OUT_OF_SCOPE",
            "missing_fields": []
        }

    required = REQUIRED_FIELDS.get(intent, [])
    missing = []

    for field in required:
        if entities.get(field) is None:
            missing.append(field)

    return {
        "status": "OK" if len(missing) == 0 else "NEEDS_INFO",
        "missing_fields": missing
    }
````

````python
tests = [
    {
        "intent": "OPEN_FOOD_BUSINESS",
        "entities": {"location": "Pune", "business_type": "cafe"},
        "location_scope": "MAHARASHTRA"
    },
    {
        "intent": "OPEN_FOOD_BUSINESS",
        "entities": {"location": None, "business_type": "cafe"},
        "location_scope": "MISSING"
    },
    {
        "intent": "ORGANIZE_PUBLIC_EVENT",
        "entities": {"location": "Delhi", "event_type": "music event"},
        "location_scope": "OUTSIDE_MAHARASHTRA"
    },
    {
        "intent": "BIRTH_CERTIFICATE",
        "entities": {"location": None},
        "location_scope": "MISSING"
    }
]

for t in tests:
    print(detect_missing_fields(
        t["intent"],
        t["entities"],
        t["location_scope"]
    ))
````

````python
CONFIDENCE_THRESHOLD = 0.65

def apply_confidence_logic(predicted_intent, confidence):
    if confidence >= CONFIDENCE_THRESHOLD:
        return {
            "use_custom_model": True,
            "needs_gemini_fallback": False,
            "intent": predicted_intent,
            "confidence": confidence
        }

    return {
        "use_custom_model": False,
        "needs_gemini_fallback": True,
        "intent": predicted_intent,
        "confidence": confidence
    }
````

````python
tests = [
    ("OPEN_FOOD_BUSINESS", 0.79),
    ("BIRTH_CERTIFICATE", 0.62),
    ("ORGANIZE_PUBLIC_EVENT", 0.71),
    ("DOMICILE_CERTIFICATE_MAHARASHTRA", 0.53)
]

for intent, conf in tests:
    print(apply_confidence_logic(intent, conf))
````

````python
def predict_with_confidence(text):
    vec = vectorizer.transform([text])

    prediction = model.predict(vec)[0]
    probabilities = model.predict_proba(vec)[0]

    confidence = probabilities.max()

    decision = apply_confidence_logic(
        prediction,
        float(confidence)
    )

    return decision
````

````python
import json

SUPPORTED_INTENTS = [
    "OPEN_FOOD_BUSINESS",
    "BIRTH_CERTIFICATE",
    "ORGANIZE_PUBLIC_EVENT",
    "DOMICILE_CERTIFICATE_MAHARASHTRA",
    "OTHER_UNSUPPORTED"
]

def gemini_intent_fallback(user_query):
    prompt = f"""
You are an intent classifier for a Maharashtra civic-services assistant.

Classify the user query into exactly ONE of these intents:

{SUPPORTED_INTENTS}

Definitions:
- OPEN_FOOD_BUSINESS: opening a restaurant, cafe, bakery, cloud kitchen, food stall, food outlet, etc. in Maharashtra.
- BIRTH_CERTIFICATE: birth registration, birth certificate copy, duplicate, correction, delayed registration.
- ORGANIZE_PUBLIC_EVENT: college fest, public gathering, cultural event, music event, sports event, exhibition, outdoor event, etc.
- DOMICILE_CERTIFICATE_MAHARASHTRA: Maharashtra domicile/residence certificate related requests.
- OTHER_UNSUPPORTED: anything else outside these supported tasks.

Return ONLY valid JSON in this exact format:
{{
  "intent": "INTENT_NAME"
}}

User query:
{user_query}
"""

    interaction = client.interactions.create(
        model="gemini-3.8-flash",
        input=prompt
    )

    raw = interaction.output_text.strip()

    # remove markdown fences if Gemini adds them
    raw = raw.replace("```json", "").replace("```", "").strip()

    data = json.loads(raw)

    return data["intent"]
````

````python
fallback_tests = [
    "mera baccha pune me born hua tha certificate chahiye",
    "domicile banana hai scholarship ke liye",
    "passport renew karna hai"
]

for q in fallback_tests:
    print(q, "->", gemini_intent_fallback(q))
````

````python
def hybrid_intent_predict(user_query):
    # Step 1: custom model prediction
    custom_result = predict_with_confidence(user_query)

    # Step 2: if confidence is high, trust custom model
    if not custom_result["needs_gemini_fallback"]:
        return {
            "final_intent": custom_result["intent"],
            "source": "CUSTOM_MODEL",
            "custom_confidence": custom_result["confidence"]
        }

    # Step 3: low confidence -> Gemini fallback
    gemini_intent = gemini_intent_fallback(user_query)

    return {
        "final_intent": gemini_intent,
        "source": "GEMINI_FALLBACK",
        "custom_confidence": custom_result["confidence"]
    }
````

````python
import re

def detect_script_type(text):
    # Devanagari Unicode range
    if re.search(r'[\u0900-\u097F]', text):
        return "DEVANAGARI"

    # Mostly Roman/Latin text
    if re.search(r'[A-Za-z]', text):
        return "ROMAN"

    return "UNKNOWN"
````

````python
script_tests = [
    "I want to open a cafe in Pune",
    "mumbai me restaurant kholna hai",
    "मुझे पुणे में डोमिसाइल चाहिए",
    "मला पुण्यात डोमिसाइल प्रमाणपत्र पाहिजे"
]

for q in script_tests:
    print(q, "->", detect_script_type(q))
````

````python
def normalize_multilingual_query(user_query):
    script_type = detect_script_type(user_query)

    # Roman text ko as-is use karenge
    if script_type == "ROMAN":
        return {
            "original_query": user_query,
            "normalized_query": user_query,
            "normalization_source": "NONE"
        }

    # Devanagari ko Gemini se simple English me normalize karenge
    prompt = f"""
Convert this Maharashtra civic-service query into simple English.

Rules:
- Preserve the exact meaning.
- Do not add any new facts.
- Keep Maharashtra city/location names if present.
- Keep the sentence short and suitable for intent classification.
- Return ONLY the translated/normalized sentence.

Query:
{user_query}
"""

    interaction = client.interactions.create(
        model="gemini-3.8-flash",
        input=prompt
    )

    normalized = interaction.output_text.strip()

    return {
        "original_query": user_query,
        "normalized_query": normalized,
        "normalization_source": "GEMINI"
    }
````

````python
multilingual_tests = [
    "I want to open a cafe in Pune",
    "mumbai me restaurant kholna hai",
    "मुझे पुणे में डोमिसाइल चाहिए",
    "मला पुण्यात डोमिसाइल प्रमाणपत्र पाहिजे"
]

for q in multilingual_tests:
    print(normalize_multilingual_query(q))
    print("-" * 60)
````

````python
def multilingual_hybrid_predict(user_query):
    normalized_data = normalize_multilingual_query(user_query)

    normalized_query = normalized_data["normalized_query"]

    intent_result = hybrid_intent_predict(normalized_query)

    return {
        "original_query": user_query,
        "normalized_query": normalized_query,
        "normalization_source": normalized_data["normalization_source"],
        "final_intent": intent_result["final_intent"],
        "intent_source": intent_result["source"],
        "custom_confidence": intent_result["custom_confidence"]
    }
````

````python
multi_predict_tests = [
    "I want to open a cafe in Pune",
    "mumbai me restaurant kholna hai",
    "मुझे पुणे में डोमिसाइल चाहिए",
    "मला पुण्यात डोमिसाइल प्रमाणपत्र पाहिजे",
    "मुझे पासपोर्ट रिन्यू करना है"
]

for q in multi_predict_tests:
    print(multilingual_hybrid_predict(q))
    print("-" * 60)
````

````python
def multilingual_full_analysis(user_query):
    normalized_data = normalize_multilingual_query(user_query)

    normalized_query = normalized_data["normalized_query"]

    intent_result = hybrid_intent_predict(normalized_query)

    final_intent = intent_result["final_intent"]

    # location + Maharashtra scope
    location_data = extract_location_with_scope(normalized_query)

    # entity extraction
    entities = extract_entities(
        normalized_query,
        final_intent
    )

    # use scope-aware location
    entities["location"] = location_data["location"]

    return {
        "original_query": user_query,
        "normalized_query": normalized_query,
        "normalization_source": normalized_data["normalization_source"],
        "final_intent": final_intent,
        "intent_source": intent_result["source"],
        "custom_confidence": intent_result["custom_confidence"],
        "location_scope": location_data["location_scope"],
        "entities": entities
    }
````

````python
KNOWN_CUSTOM_INTENTS = {
    "OPEN_FOOD_BUSINESS",
    "BIRTH_CERTIFICATE",
    "ORGANIZE_PUBLIC_EVENT",
    "DOMICILE_CERTIFICATE_MAHARASHTRA"
}

def normalize_intent_route(predicted_intent):
    if predicted_intent in KNOWN_CUSTOM_INTENTS:
        return {
            "intent": predicted_intent,
            "route": "CUSTOM_FAST_PATH"
        }

    # Existing OTHER_UNSUPPORTED model class
    # actually represents other civic-service queries.
    if predicted_intent == "OTHER_UNSUPPORTED":
        return {
            "intent": "GENERAL_CIVIC_TASK",
            "route": "GENERIC_CIVIC_PATH"
        }

    return {
        "intent": "GENERAL_CIVIC_TASK",
        "route": "GENERIC_CIVIC_PATH"
    }
````

````python
def build_final_response(
    query,
    normalized_query,
    intent,
    route,
    location,
    location_scope,
    entities,
    confidence,
    intent_source,
    status,
    missing_fields
):
    return {
        "query": query,
        "normalized_query": normalized_query,
        "intent": intent,
        "route": route,
        "location": location,
        "location_scope": location_scope,
        "entities": entities,
        "confidence": confidence,
        "intent_source": intent_source,
        "status": status,
        "missing_fields": missing_fields
    }

print("Final response builder ready")
````

````python
def analyze_query_core(user_query):
    # 1. Custom model prediction + confidence
    custom_result = predict_with_confidence(user_query)

    # 2. Route normalization
    route_info = normalize_intent_route(custom_result["intent"])

    final_intent = route_info["intent"]
    route = route_info["route"]

    # 3. Location + scope
    location_data = extract_location_with_scope(user_query)

    # 4. Entity extraction
    entities = extract_entities(
        user_query,
        custom_result["intent"]
    )

    entities["location"] = location_data["location"]

    # 5. Missing-field detection
    missing_result = detect_missing_fields(
        custom_result["intent"],
        entities,
        location_data["location_scope"]
    )

    # 6. Final standardized response
    return build_final_response(
        query=user_query,
        normalized_query=user_query,
        intent=final_intent,
        route=route,
        location=location_data["location"],
        location_scope=location_data["location_scope"],
        entities=entities,
        confidence=float(custom_result["confidence"]),
        intent_source="CUSTOM_MODEL",
        status=missing_result["status"],
        missing_fields=missing_result["missing_fields"]
    )

print("Core analyzer ready")
````

````python
result = analyze_query_core(
    "I want to open a cafe in Pune"
)

print(result)
````

````python
def predict_with_confidence(text):
    vec = vectorizer.transform([text])

    prediction = model.predict(vec)[0]
    probabilities = model.predict_proba(vec)[0]

    confidence = probabilities.max()

    decision = apply_confidence_logic(
        prediction,
        float(confidence)
    )

    return decision

print("predict_with_confidence ready")
````

````python
CONFIDENCE_THRESHOLD = 0.65

def apply_confidence_logic(predicted_intent, confidence):
    if confidence >= CONFIDENCE_THRESHOLD:
        return {
            "use_custom_model": True,
            "needs_gemini_fallback": False,
            "intent": predicted_intent,
            "confidence": confidence
        }

    return {
        "use_custom_model": False,
        "needs_gemini_fallback": True,
        "intent": predicted_intent,
        "confidence": confidence
    }

print("apply_confidence_logic ready")
````

````python
MAHARASHTRA_CITIES = [
    "mumbai",
    "pune",
    "thane",
    "navi mumbai",
    "nagpur",
    "nashik",
    "kolhapur",
    "solapur",
    "amravati",
    "kalyan",
    "dombivli",
    "chhatrapati sambhajinagar"
]

OUTSIDE_MAHARASHTRA_CITIES = [
    "delhi",
    "new delhi",
    "bangalore",
    "bengaluru",
    "hyderabad",
    "chennai",
    "kolkata",
    "ahmedabad",
    "jaipur",
    "lucknow",
    "surat",
    "indore",
    "bhopal",
    "chandigarh",
    "patna",
    "ranchi",
    "kochi",
    "gurgaon",
    "gurugram",
    "noida"
]

FOOD_BUSINESS_TYPES = [
    "restaurant",
    "bakery",
    "cafe",
    "cloud kitchen",
    "food stall",
    "juice centre",
    "juice shop",
    "snack centre",
    "eatery",
    "food outlet",
    "tiffin service",
    "catering business",
    "sweet shop",
    "fast food shop",
    "tea cafe",
    "home kitchen"
]

EVENT_TYPES = [
    "college fest",
    "cultural program",
    "music event",
    "community gathering",
    "outdoor festival",
    "sports event",
    "public exhibition",
    "stage show",
    "charity event",
    "community function",
    "public celebration",
    "college cultural event",
    "outdoor function",
    "public performance",
    "public event"
]

BIRTH_SUBTYPES = {
    "new_registration": [
        "newborn",
        "new birth",
        "birth registration",
        "register birth",
        "register the birth",
        "baby born",
        "child born"
    ],
    "duplicate_copy": [
        "duplicate",
        "copy",
        "reissue",
        "old birth certificate"
    ],
    "correction": [
        "correction",
        "spelling mistake",
        "name correction",
        "date correction",
        "wrong name",
        "wrong date"
    ],
    "delayed_registration": [
        "delayed",
        "not registered",
        "never registered",
        "late registration"
    ]
}

DOMICILE_PURPOSES = {
    "admission": [
        "admission",
        "college",
        "university",
        "education"
    ],
    "scholarship": [
        "scholarship"
    ],
    "government_job": [
        "government job",
        "govt job",
        "job application"
    ],
    "state_benefit": [
        "state benefit",
        "government scheme",
        "scheme"
    ]
}

def extract_location_with_scope(text):
    text_lower = text.lower()

    for city in MAHARASHTRA_CITIES:
        if city in text_lower:
            return {
                "location": city.title(),
                "location_scope": "MAHARASHTRA"
            }

    for city in OUTSIDE_MAHARASHTRA_CITIES:
        if city in text_lower:
            return {
                "location": city.title(),
                "location_scope": "OUTSIDE_MAHARASHTRA"
            }

    return {
        "location": None,
        "location_scope": "MISSING"
    }

def extract_business_type(text):
    text_lower = text.lower()

    for business in FOOD_BUSINESS_TYPES:
        if business in text_lower:
            return business

    return None

def extract_event_type(text):
    text_lower = text.lower()

    for event in EVENT_TYPES:
        if event in text_lower:
            return event

    return None

def extract_birth_subtype(text):
    text_lower = text.lower()

    for subtype, keywords in BIRTH_SUBTYPES.items():
        for keyword in keywords:
            if keyword in text_lower:
                return subtype

    return None

def extract_domicile_purpose(text):
    text_lower = text.lower()

    for purpose, keywords in DOMICILE_PURPOSES.items():
        for keyword in keywords:
            if keyword in text_lower:
                return purpose

    return None

def extract_entities(text, predicted_intent):
    result = {
        "location": None,
        "business_type": None,
        "event_type": None,
        "birth_subtype": None,
        "domicile_purpose": None
    }

    if predicted_intent == "OPEN_FOOD_BUSINESS":
        result["business_type"] = extract_business_type(text)

    elif predicted_intent == "ORGANIZE_PUBLIC_EVENT":
        result["event_type"] = extract_event_type(text)

    elif predicted_intent == "BIRTH_CERTIFICATE":
        result["birth_subtype"] = extract_birth_subtype(text)

    elif predicted_intent == "DOMICILE_CERTIFICATE_MAHARASHTRA":
        result["domicile_purpose"] = extract_domicile_purpose(text)

    return result

REQUIRED_FIELDS = {
    "OPEN_FOOD_BUSINESS": [
        "location",
        "business_type"
    ],
    "ORGANIZE_PUBLIC_EVENT": [
        "location",
        "event_type"
    ],
    "BIRTH_CERTIFICATE": [
        "location"
    ],
    "DOMICILE_CERTIFICATE_MAHARASHTRA": [],
    "OTHER_UNSUPPORTED": []
}

def detect_missing_fields(intent, entities, location_scope):
    if location_scope == "OUTSIDE_MAHARASHTRA":
        return {
            "status": "OUT_OF_SCOPE",
            "missing_fields": []
        }

    required = REQUIRED_FIELDS.get(intent, [])
    missing = []

    for field in required:
        if entities.get(field) is None:
            missing.append(field)

    return {
        "status": "OK" if len(missing) == 0 else "NEEDS_INFO",
        "missing_fields": missing
    }

print("Entity, scope and missing-field helpers restored")
````

````python
import os
import json
import copy

GOV_DATA_DIR = "/content/drive/MyDrive/PSWB02_AI/government_data"

tasks = {
    "BIRTH_CERTIFICATE": "birth_certificate.json",
    "DEATH_CERTIFICATE": "death_certificate.json",
    "MARRIAGE_REGISTRATION": "marriage_registration.json",
    "DOMICILE_CERTIFICATE": "domicile_certificate.json",
    "INCOME_CERTIFICATE": "income_certificate.json",
    "CASTE_CERTIFICATE": "caste_certificate.json",
    "NON_CREAMY_LAYER_CERTIFICATE": "non_creamy_layer_certificate.json",
    "NEW_RATION_CARD": "new_ration_card.json",
    "RATION_CARD_UPDATE": "ration_card_update.json",
    "SHOP_ESTABLISHMENT_REGISTRATION": "shop_establishment_registration.json",
    "OPEN_FOOD_BUSINESS": "open_food_business.json",
    "PROPERTY_CIVIC_SERVICE": "property_civic_service.json",
    "NEW_WATER_CONNECTION": "new_water_connection.json",
    "TRADE_LICENCE": "trade_licence.json",
    "HEALTH_LICENCE": "health_licence.json",
    "BUILDING_PERMIT": "building_permit.json",
    "PUBLIC_EVENT_PERMISSION": "public_event_permission.json",
    "ASSEMBLY_PROCESSION_PERMISSION": "assembly_procession_permission.json",
    "SENIOR_CITIZEN_CERTIFICATE": "senior_citizen_certificate.json",
    "POLICE_CLEARANCE": "police_clearance.json"
}

base_schema = {
    "task_id": "",
    "task_name": "",
    "category": "",
    "jurisdiction": {
        "state": "Maharashtra",
        "district": None,
        "local_body": None
    },
    "department": {
        "name": "",
        "sub_department": None,
        "designated_officer": None,
        "office_type": None
    },
    "eligibility": [],
    "steps": [],
    "status_options": [
        "NOT_STARTED",
        "IN_PROGRESS",
        "COMPLETED",
        "LOCKED"
    ],
    "source_freshness": {
        "last_verified": "",
        "needs_review": True
    }
}

# 1. Create 20 detailed task JSON files
for task_id, filename in tasks.items():
    data = copy.deepcopy(base_schema)
    data["task_id"] = task_id
    data["task_name"] = task_id.replace("_", " ").title()

    path = os.path.join(GOV_DATA_DIR, filename)

    with open(path, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2, ensure_ascii=False)

# 2. Create task_catalog.json
task_catalog = [
    {
        "task_id": task_id,
        "task_name": task_id.replace("_", " ").title(),
        "state": "Maharashtra",
        "data_file": filename
    }
    for task_id, filename in tasks.items()
]

with open(
    os.path.join(GOV_DATA_DIR, "task_catalog.json"),
    "w",
    encoding="utf-8"
) as f:
    json.dump(task_catalog, f, indent=2, ensure_ascii=False)

# 3. Create task_mapping.json
task_mapping = {
    **tasks,
    "GENERAL_CIVIC_TASK": "dynamic_retrieval"
}

with open(
    os.path.join(GOV_DATA_DIR, "task_mapping.json"),
    "w",
    encoding="utf-8"
) as f:
    json.dump(task_mapping, f, indent=2, ensure_ascii=False)

# 4. Verify
files = sorted(os.listdir(GOV_DATA_DIR))

print("Total JSON files:", len(files))

for file in files:
    print(file)
````

````python
import os
import json

GOV_DATA_DIR = "/content/drive/MyDrive/PSWB02_AI/government_data"

expected_task_files = [
    "assembly_procession_permission.json",
    "birth_certificate.json",
    "building_permit.json",
    "caste_certificate.json",
    "death_certificate.json",
    "domicile_certificate.json",
    "health_licence.json",
    "income_certificate.json",
    "marriage_registration.json",
    "new_ration_card.json",
    "new_water_connection.json",
    "non_creamy_layer_certificate.json",
    "open_food_business.json",
    "police_clearance.json",
    "property_civic_service.json",
    "public_event_permission.json",
    "ration_card_update.json",
    "senior_citizen_certificate.json",
    "shop_establishment_registration.json",
    "trade_licence.json"
]

required_top_keys = [
    "task_id",
    "task_name",
    "category",
    "jurisdiction",
    "department",
    "eligibility",
    "steps",
    "status_options",
    "source_freshness"
]

required_step_keys = [
    "step_id",
    "title",
    "description",
    "required_forms",
    "required_documents",
    "fees",
    "office",
    "prerequisites",
    "depends_on",
    "unlocks",
    "can_run_in_parallel",
    "application",
    "time_limit_days",
    "official_sources"
]

errors = []
warnings = []

print("=" * 80)
print("PSWB02 GOVERNMENT DATA VALIDATION")
print("=" * 80)

# -------------------------------------------------------
# 1. CHECK FILE COUNT
# -------------------------------------------------------

json_files = sorted([
    f for f in os.listdir(GOV_DATA_DIR)
    if f.endswith(".json")
])

print("\nTotal JSON files found:", len(json_files))

if len(json_files) != 22:
    errors.append(
        f"Expected 22 JSON files but found {len(json_files)}"
    )

# -------------------------------------------------------
# 2. CHECK ALL 20 DETAILED TASK FILES EXIST
# -------------------------------------------------------

for filename in expected_task_files:
    path = os.path.join(GOV_DATA_DIR, filename)

    if not os.path.exists(path):
        errors.append(f"Missing task file: {filename}")

# -------------------------------------------------------
# 3. VALIDATE EACH DETAILED TASK JSON
# -------------------------------------------------------

validated_tasks = []

for filename in expected_task_files:

    path = os.path.join(GOV_DATA_DIR, filename)

    if not os.path.exists(path):
        continue

    try:
        with open(path, "r", encoding="utf-8") as f:
            data = json.load(f)

    except Exception as e:
        errors.append(
            f"{filename}: invalid JSON -> {e}"
        )
        continue

    # Top-level keys
    for key in required_top_keys:
        if key not in data:
            errors.append(
                f"{filename}: missing top-level key '{key}'"
            )

    # Task ID
    if not data.get("task_id"):
        errors.append(
            f"{filename}: empty task_id"
        )

    # Steps
    steps = data.get("steps", [])

    if not isinstance(steps, list) or len(steps) == 0:
        errors.append(
            f"{filename}: no valid steps"
        )
        continue

    step_ids = []

    for i, step in enumerate(steps, start=1):

        for key in required_step_keys:
            if key not in step:
                errors.append(
                    f"{filename} -> step {i}: missing '{key}'"
                )

        step_id = step.get("step_id")

        if step_id:
            step_ids.append(step_id)

        sources = step.get("official_sources", [])

        if not sources:
            warnings.append(
                f"{filename} -> {step_id}: no official source"
            )

        for source in sources:
            url = source.get("source_url")

            if not url:
                warnings.append(
                    f"{filename} -> {step_id}: source missing URL"
                )

            if source.get("source_type") != "official_government":
                warnings.append(
                    f"{filename} -> {step_id}: "
                    f"source_type is not official_government"
                )

    # Duplicate step IDs
    if len(step_ids) != len(set(step_ids)):
        errors.append(
            f"{filename}: duplicate step_id detected"
        )

    # Dependency references
    valid_step_ids = set(step_ids)

    for step in steps:

        for dependency in step.get("depends_on", []):
            if dependency not in valid_step_ids:
                warnings.append(
                    f"{filename} -> {step['step_id']}: "
                    f"depends_on references unknown step '{dependency}'"
                )

        for unlock in step.get("unlocks", []):
            if unlock not in valid_step_ids:
                warnings.append(
                    f"{filename} -> {step['step_id']}: "
                    f"unlocks references unknown step '{unlock}'"
                )

    validated_tasks.append(
        data.get("task_id")
    )

    print(
        f"✅ {filename:<42} "
        f"steps={len(steps):<2} "
        f"task_id={data.get('task_id')}"
    )

# -------------------------------------------------------
# 4. VALIDATE TASK CATALOG
# -------------------------------------------------------

catalog_path = os.path.join(
    GOV_DATA_DIR,
    "task_catalog.json"
)

if not os.path.exists(catalog_path):

    errors.append("task_catalog.json missing")

else:

    try:
        with open(catalog_path, "r", encoding="utf-8") as f:
            catalog = json.load(f)

        print("\ntask_catalog entries:", len(catalog))

        if len(catalog) != 20:
            errors.append(
                f"task_catalog should contain 20 entries, found {len(catalog)}"
            )

        for entry in catalog:
            data_file = entry.get("data_file")

            if not data_file:
                errors.append(
                    f"task_catalog entry missing data_file: {entry}"
                )

            elif not os.path.exists(
                os.path.join(GOV_DATA_DIR, data_file)
            ):
                errors.append(
                    f"task_catalog points to missing file: {data_file}"
                )

    except Exception as e:
        errors.append(
            f"task_catalog.json invalid: {e}"
        )

# -------------------------------------------------------
# 5. VALIDATE TASK MAPPING
# -------------------------------------------------------

mapping_path = os.path.join(
    GOV_DATA_DIR,
    "task_mapping.json"
)

if not os.path.exists(mapping_path):

    errors.append("task_mapping.json missing")

else:

    try:
        with open(mapping_path, "r", encoding="utf-8") as f:
            mapping = json.load(f)

        print("task_mapping entries:", len(mapping))

        for task_id, target in mapping.items():

            if task_id == "GENERAL_CIVIC_TASK":
                if target != "dynamic_retrieval":
                    warnings.append(
                        "GENERAL_CIVIC_TASK should map to dynamic_retrieval"
                    )
                continue

            if not os.path.exists(
                os.path.join(GOV_DATA_DIR, target)
            ):
                errors.append(
                    f"task_mapping: {task_id} -> missing file {target}"
                )

    except Exception as e:
        errors.append(
            f"task_mapping.json invalid: {e}"
        )

# -------------------------------------------------------
# 6. FINAL RESULT
# -------------------------------------------------------

print("\n" + "=" * 80)

if not errors:
    print("✅ VALIDATION PASSED")
else:
    print("❌ VALIDATION FAILED")

print("=" * 80)

print("\nErrors:", len(errors))
for e in errors:
    print("❌", e)

print("\nWarnings:", len(warnings))
for w in warnings:
    print("⚠️", w)

print("\nValidated detailed tasks:", len(validated_tasks))
print("Total JSON files:", len(json_files))
````

````python
from pathlib import Path

APP_PATH = Path("/content/drive/MyDrive/PSWB02_AI/app.py")

app_code = r'''
from pathlib import Path
import json
import re
import joblib

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel


# ============================================================
# PATHS
# ============================================================

BASE_DIR = Path(__file__).resolve().parent

MODEL_PATH = BASE_DIR / "models" / "intent_classifier.joblib"
VECTORIZER_PATH = BASE_DIR / "models" / "tfidf_vectorizer.joblib"
GOV_DATA_DIR = BASE_DIR / "government_data"


# ============================================================
# LOAD MODEL
# ============================================================

if not MODEL_PATH.exists():
    raise FileNotFoundError(f"Intent model not found: {MODEL_PATH}")

if not VECTORIZER_PATH.exists():
    raise FileNotFoundError(f"Vectorizer not found: {VECTORIZER_PATH}")

intent_model = joblib.load(MODEL_PATH)
tfidf_vectorizer = joblib.load(VECTORIZER_PATH)


# ============================================================
# CONFIG
# ============================================================

CONFIDENCE_THRESHOLD = 0.65

KNOWN_CUSTOM_INTENTS = {
    "OPEN_FOOD_BUSINESS",
    "BIRTH_CERTIFICATE",
    "ORGANIZE_PUBLIC_EVENT",
    "DOMICILE_CERTIFICATE_MAHARASHTRA"
}

MAHARASHTRA_CITIES = [
    "Mumbai",
    "Pune",
    "Thane",
    "Navi Mumbai",
    "Nagpur",
    "Nashik",
    "Kolhapur",
    "Solapur",
    "Amravati",
    "Kalyan",
    "Dombivli",
    "Chhatrapati Sambhajinagar"
]

OUTSIDE_MAHARASHTRA_CITIES = [
    "Delhi",
    "New Delhi",
    "Bangalore",
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Kolkata",
    "Ahmedabad",
    "Jaipur",
    "Lucknow",
    "Surat",
    "Indore",
    "Bhopal",
    "Chandigarh",
    "Patna",
    "Ranchi",
    "Kochi",
    "Gurgaon",
    "Gurugram",
    "Noida"
]

FOOD_BUSINESS_TYPES = [
    "restaurant",
    "bakery",
    "cafe",
    "cloud kitchen",
    "food stall",
    "juice centre",
    "juice center",
    "juice shop",
    "snack centre",
    "snack center",
    "eatery",
    "food outlet",
    "tiffin service",
    "catering business",
    "sweet shop",
    "fast food shop",
    "tea cafe",
    "home kitchen"
]

EVENT_TYPES = [
    "college fest",
    "cultural program",
    "music event",
    "community gathering",
    "outdoor festival",
    "sports event",
    "public exhibition",
    "stage show",
    "charity event",
    "community function",
    "public celebration",
    "college cultural event",
    "outdoor function",
    "public performance",
    "public event"
]

REQUIRED_FIELDS = {
    "OPEN_FOOD_BUSINESS": ["location", "business_type"],
    "ORGANIZE_PUBLIC_EVENT": ["location", "event_type"],
    "BIRTH_CERTIFICATE": ["location"],
    "DOMICILE_CERTIFICATE_MAHARASHTRA": []
}


# ============================================================
# REQUEST MODEL
# ============================================================

class QueryRequest(BaseModel):
    query: str


# ============================================================
# HELPERS
# ============================================================

def find_phrase(text, phrases):
    text_lower = text.lower()

    for phrase in sorted(phrases, key=len, reverse=True):
        if phrase.lower() in text_lower:
            return phrase

    return None


def extract_location_with_scope(text):
    text_lower = text.lower()

    for city in sorted(MAHARASHTRA_CITIES, key=len, reverse=True):
        if city.lower() in text_lower:
            return city, "MAHARASHTRA"

    for city in sorted(
        OUTSIDE_MAHARASHTRA_CITIES,
        key=len,
        reverse=True
    ):
        if city.lower() in text_lower:
            return city, "OUTSIDE_MAHARASHTRA"

    return None, "MISSING"


def extract_birth_subtype(text):
    t = text.lower()

    if any(x in t for x in [
        "delayed",
        "not registered",
        "never registered",
        "late registration"
    ]):
        return "delayed_registration"

    if any(x in t for x in [
        "correction",
        "spelling mistake",
        "name correction",
        "date correction",
        "wrong name",
        "wrong date"
    ]):
        return "correction"

    if any(x in t for x in [
        "duplicate",
        "copy",
        "reissue",
        "old birth certificate"
    ]):
        return "duplicate_copy"

    if any(x in t for x in [
        "newborn",
        "new birth",
        "birth registration",
        "register birth",
        "register the birth",
        "baby born",
        "child born"
    ]):
        return "new_registration"

    return None


def extract_domicile_purpose(text):
    t = text.lower()

    if any(x in t for x in [
        "admission",
        "college",
        "university",
        "education"
    ]):
        return "admission"

    if "scholarship" in t:
        return "scholarship"

    if any(x in t for x in [
        "government job",
        "govt job",
        "job application"
    ]):
        return "government_job"

    if any(x in t for x in [
        "state benefit",
        "government scheme",
        "govt scheme",
        "scheme"
    ]):
        return "state_benefit"

    return None


def extract_entities(text, predicted_intent):
    location, _ = extract_location_with_scope(text)

    entities = {
        "location": location,
        "business_type": None,
        "event_type": None,
        "birth_subtype": None,
        "domicile_purpose": None
    }

    if predicted_intent == "OPEN_FOOD_BUSINESS":
        entities["business_type"] = find_phrase(
            text,
            FOOD_BUSINESS_TYPES
        )

    elif predicted_intent == "ORGANIZE_PUBLIC_EVENT":
        entities["event_type"] = find_phrase(
            text,
            EVENT_TYPES
        )

    elif predicted_intent == "BIRTH_CERTIFICATE":
        entities["birth_subtype"] = extract_birth_subtype(text)

    elif predicted_intent == "DOMICILE_CERTIFICATE_MAHARASHTRA":
        entities["domicile_purpose"] = extract_domicile_purpose(text)

    return entities


def predict_intent(text):
    vector = tfidf_vectorizer.transform([text])

    predicted_intent = intent_model.predict(vector)[0]

    probabilities = intent_model.predict_proba(vector)[0]

    confidence = float(max(probabilities))

    return predicted_intent, confidence


def normalize_intent_route(predicted_intent):
    if predicted_intent in KNOWN_CUSTOM_INTENTS:
        return {
            "intent": predicted_intent,
            "route": "CUSTOM_FAST_PATH"
        }

    return {
        "intent": "GENERAL_CIVIC_TASK",
        "route": "GENERIC_CIVIC_PATH"
    }


def detect_missing_fields(intent, entities, location_scope):
    if location_scope == "OUTSIDE_MAHARASHTRA":
        return "OUT_OF_SCOPE", []

    required = REQUIRED_FIELDS.get(intent, [])

    missing = [
        field
        for field in required
        if not entities.get(field)
    ]

    if missing:
        return "NEEDS_INFO", missing

    return "OK", []


def analyze_query_core(user_query):
    if not user_query or not user_query.strip():
        raise ValueError("Query cannot be empty.")

    normalized_query = user_query.strip()

    raw_intent, confidence = predict_intent(normalized_query)

    route_info = normalize_intent_route(raw_intent)

    final_intent = route_info["intent"]
    route = route_info["route"]

    location, location_scope = extract_location_with_scope(
        normalized_query
    )

    entity_intent = (
        raw_intent
        if raw_intent in KNOWN_CUSTOM_INTENTS
        else final_intent
    )

    entities = extract_entities(
        normalized_query,
        entity_intent
    )

    entities["location"] = location

    if route == "CUSTOM_FAST_PATH":
        status, missing_fields = detect_missing_fields(
            raw_intent,
            entities,
            location_scope
        )
    else:
        if location_scope == "OUTSIDE_MAHARASHTRA":
            status = "OUT_OF_SCOPE"
        else:
            status = "OK"

        missing_fields = []

    intent_source = (
        "CUSTOM_MODEL"
        if confidence >= CONFIDENCE_THRESHOLD
        else "CUSTOM_MODEL_LOW_CONFIDENCE"
    )

    return {
        "query": user_query,
        "normalized_query": normalized_query,
        "intent": final_intent,
        "route": route,
        "location": location,
        "location_scope": location_scope,
        "entities": entities,
        "confidence": confidence,
        "intent_source": intent_source,
        "status": status,
        "missing_fields": missing_fields
    }


def load_task_mapping():
    mapping_path = GOV_DATA_DIR / "task_mapping.json"

    if not mapping_path.exists():
        raise FileNotFoundError(
            f"task_mapping.json not found: {mapping_path}"
        )

    with open(mapping_path, "r", encoding="utf-8") as f:
        return json.load(f)


def get_task_data(task_id):
    mapping = load_task_mapping()

    if task_id not in mapping:
        raise KeyError(task_id)

    target = mapping[task_id]

    if target == "dynamic_retrieval":
        return {
            "task_id": task_id,
            "route": "dynamic_retrieval"
        }

    task_path = GOV_DATA_DIR / target

    if not task_path.exists():
        raise FileNotFoundError(
            f"Government data file not found: {task_path}"
        )

    with open(task_path, "r", encoding="utf-8") as f:
        return json.load(f)


# ============================================================
# FASTAPI
# ============================================================

app = FastAPI(
    title="PSWB02 Civic Task Navigator AI",
    version="1.0.0",
    description=(
        "AI service for Maharashtra civic-task intent routing, "
        "entity extraction and verified government workflow retrieval."
    )
)


@app.get("/")
def root():
    return {
        "service": "PSWB02 Civic Task Navigator AI",
        "status": "running"
    }


@app.get("/health")
def health():
    return {
        "status": "ok",
        "model_loaded": True,
        "government_data_directory": GOV_DATA_DIR.exists()
    }


@app.post("/ai/analyze")
def analyze(request: QueryRequest):
    try:
        return analyze_query_core(request.query)

    except Exception as e:
        raise HTTPException(
            status_code=500,
            detail=str(e)
        )


@app.get("/government/task/{task_id}")
def government_task(task_id: str):
    task_id = task_id.upper()

    try:
        return get_task_data(task_id)

    except KeyError:
        raise HTTPException(
            status_code=404,
            detail=f"Unknown task_id: {task_id}"
        )

    except Exception as e:
        raise HTTPException(
            status_code=500,
            detail=str(e)
        )
'''

APP_PATH.write_text(app_code, encoding="utf-8")

print("Created:", APP_PATH)
print("Size:", APP_PATH.stat().st_size, "bytes")
````

````python
from pathlib import Path

README_PATH = Path("/content/drive/MyDrive/PSWB02_AI/README.md")

readme_text = '''
# PSWB02 - Municipal Bureaucracy Path Visualizer

AI and government-data module for a Civic Task Navigator focused on Maharashtra civic services.

## What this module does

- Accepts natural-language civic queries
- Predicts civic intent using TF-IDF + Logistic Regression
- Extracts location and task-specific entities
- Detects missing information
- Routes supported tasks through a custom fast path
- Routes other civic tasks through a generic civic path
- Loads structured government workflow JSON files
- Exposes FastAPI endpoints for backend integration

## Current custom intents

- OPEN_FOOD_BUSINESS
- BIRTH_CERTIFICATE
- ORGANIZE_PUBLIC_EVENT
- DOMICILE_CERTIFICATE_MAHARASHTRA

Other civic queries are routed to:

- GENERAL_CIVIC_TASK
- GENERIC_CIVIC_PATH

## Government data

The government_data folder contains:

- 20 detailed Maharashtra civic-task JSON files
- task_catalog.json
- task_mapping.json

Total: 22 JSON files

Each detailed task JSON can contain:

- department
- designated officer
- eligibility
- forms
- required documents
- fees
- prerequisites
- dependencies
- application links
- processing timelines
- official government sources
- progress states

Government facts are sourced from official government portals. Source URLs are stored inside the relevant workflow steps.

## API endpoints

GET /health

Checks whether the AI service and government data directory are available.

POST /ai/analyze

Example request:

{
    "query": "I want to open a cafe in Pune"
}

Main response fields:

- query
- normalized_query
- intent
- route
- location
- location_scope
- entities
- confidence
- intent_source
- status
- missing_fields

GET /government/task/{task_id}

Example:

GET /government/task/DOMICILE_CERTIFICATE

## Project structure

PSWB02_AI/
- app.py
- README.md
- requirements.txt
- .gitignore
- intent_model.ipynb
- data/
- models/
  - intent_classifier.joblib
  - tfidf_vectorizer.joblib
- government_data/
  - task_catalog.json
  - task_mapping.json
  - 20 detailed task JSON files

## Run locally

Install dependencies:

pip install -r requirements.txt

Start FastAPI:

uvicorn app:app --reload

Swagger API documentation:

http://127.0.0.1:8000/docs

## Notes

- Current implementation is Maharashtra-focused.
- Municipal requirements may vary by local authority.
- When official sources conflict or do not provide a universal value, the dataset preserves that uncertainty instead of inventing information.
- Gemini and multilingual experiments remain in the notebook and are not required for the standalone local API.
'''

README_PATH.write_text(readme_text.strip() + "\n", encoding="utf-8")

print("Created:", README_PATH)
print("README size:", README_PATH.stat().st_size, "bytes")
````
